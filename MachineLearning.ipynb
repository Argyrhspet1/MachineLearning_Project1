#Convert DataFrame to Numpy

```python
x=data.drop(columns=['Outcome'])
y=data['Outcome']

x=x.to_numpy()
y=y.to_numpy()
x.shape, y.shape
```

#Split Dataset to Train-Test

Train samples: 537 , Test samples: 231

```python
seed=0
test_size=0.3

X_train, X_test, y_train, y_test=train_test_split(x,y,test_size=test_size,random_state=seed)

X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

#Create a Logistic Regression Model

```python
model=LogisticRegression(max_iter=5000,random_state=seed)
model.fit(X_train, y_train)
```

#Evaluate Model on Training Set

```python
y_pred=model.predict(X_train)
y_pred
```

```python
accuracy=accuracy_score(y_train,y_pred)
print(f"Accuracy: {round(accuracy*100,2)}%")
```

#Evaluate Model in Test Set

```python
y_preds2=model.predict(X_test)
y_preds2
```

```python
accuracy2=accuracy_score(y_test,y_preds2)
print(f"Accuracy: {round(accuracy2*100,2)}%")
```

The Accuracy in the Evaluation of the Train Dataset was slightly lower than that of the Test Dataset.
